```python
import pandas as pd

df = pd.read_csv("club_signups.csv")

df.head(10)
```

```python
print("Shape:", df.shape)
```

```python
df.dtypes
```

```python
categorical_columns = [
    "full_name",
    "email",
    "faculty",
    "city",
    "club",
    "fee_paid"
]

for column in categorical_columns:
    print(f"\n===== {column} =====")
    print(df[column].value_counts(dropna=False))
```

```python
for column in ["faculty", "city", "club"]:
    df[column] = df[column].str.strip().str.lower()
```

```python
faculty_map = {
    "met": "MET",
    "iet": "IET",
    "ems": "EMS",
    "pharmacy": "Pharmacy",
    "management": "Management",
    "information engineering and technology": "IET",
    "engineering and materials science": "EMS",
    "pharma": "Pharmacy",
    "media engineering and technology": "MET"
}

df["faculty"] = df["faculty"].replace(faculty_map)
```

```python
city_map = {
    "cairo": "Cairo",
    "giza": "Giza",
    "el giza": "Giza",
    "alexandria": "Alexandria",
    "alex": "Alexandria"
}

df["city"] = df["city"].replace(city_map)
```

```python
club_map = {
    "robotics": "Robotics",
    "debate": "Debate",
    "debating": "Debate",
    "debate club": "Debate",
    "music": "Music",
    "football": "Football",
    "soccer": "Football",
    "chess": "Chess",
    "chess club": "Chess"
}

df["club"] = df["club"].replace(club_map)
```

```python
assert set(df["faculty"].unique()) <= {
    "MET", "IET", "EMS", "Pharmacy", "Management"
}

assert set(df["city"].unique()) <= {
    "Cairo", "Giza", "Alexandria"
}

assert set(df["club"].unique()) <= {
    "Robotics", "Debate", "Music", "Football", "Chess"
}
```

```python
df["full_name"] = (
    df["full_name"]
    .str.split()
    .str.join(" ")
    .str.title()
)
```

```python
df["email"] = df["email"].str.strip().str.lower()
```

```python
df[["full_name", "email"]].head(10)
```

```python
fee_map = {
    "yes": True,
    "y": True,
    "true": True,
    "1": True,
    "no": False,
    "n": False,
    "false": False,
    "0": False
}

df["fee_paid"] = (
    df["fee_paid"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map(fee_map)
)
```

```python
print(df["fee_paid"].value_counts())
print(df["fee_paid"].dtype)
```

```python
assert df["fee_paid"].dtype == bool
```

```python
df["signed_up_at"] = pd.to_datetime(
    df["signed_up_at"],
    format="mixed",
    dayfirst=True
)
```

```python
df["signed_up_at"].dtype
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates().copy()
```

```python
print(df.shape)
```

```python
df.duplicated(subset=["student_id", "club"]).sum()
```

```python
df = df.sort_values("signed_up_at")
```

```python
df = df.drop_duplicates(
    subset=["student_id", "club"],
    keep="last"
).copy()
```

```python
assert not df.duplicated(
    subset=["student_id", "club"]
).any()
```

```python
df.to_csv("club_signups_clean.csv", index=False)
```
